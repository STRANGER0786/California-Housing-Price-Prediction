Problem Statement:

The objective of this project is to develop a predictive model for housing prices using machine learning techniques. Multiple Linear Regression is used to estimate house prices based on various housing features, while Logistic Regression is applied to classify houses into different price categories (e.g., high-priced or low-priced). The project aims to analyze the relationship between housing attributes and property values, enabling accurate price prediction and classification.

```python
import pandas as pd
import numpy as np
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
```

```python
housing = fetch_california_housing(as_frame=True)
df = housing.frame


print(df.head())
```

```python
X = df.drop("MedHouseVal",axis=1)
y = df["MedHouseVal"]

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3,random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
r2 = r2_score(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mae = mean_absolute_error(y_test, y_pred)


print("R^2 Score:", r2)
print("RMSE:", rmse)
print("MAE:", mae)
```

```python
print("Intercept:", model.intercept_)


for feature, coef in zip(X.columns, model.coef_):
    print(feature, ":", coef)
```

```python
from sklearn.datasets import fetch_california_housing
import pandas as pd

housing = fetch_california_housing(as_frame=True)
df = housing.frame
```

```python
median_price = df['MedHouseVal'].median()

df['PriceCategory'] = (df['MedHouseVal']>median_price).astype(int)
```

```python
X = df.drop(['MedHouseVal','PriceCategory'], axis=1)
y = df['PriceCategory']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=5000)

model.fit(X_train_scaled, y_train)
```

```python
y_pred = model.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))
```

Limitations

Logistic Regression can only classify houses into predefined categories and cannot predict exact prices
